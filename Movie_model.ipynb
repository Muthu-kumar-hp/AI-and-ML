```python
import pandas as pd
movie_info = pd.read_csv(r"C:\Users\muthu\Downloads\Movie Interests.csv")
```

```python
movie_info
```

```python
input_dataset = movie_info.drop(columns=['Interest'])
```

```python
input_dataset
```

```python
output_dataset = movie_info['Interest']
```

```python
output_dataset
```

```python
from sklearn.tree import DecisionTreeClassifier
movie_model = DecisionTreeClassifier()
movie_model.fit(input_dataset,output_dataset)
movie_interset= movie_model.predict([[9,2],[35,0]])
```

```python
movie_interset
```

```python
from sklearn.neighbors import KNeighborsClassifier
movie_model = KNeighborsClassifier(n_neighbors=3)
movie_model.fit(input_dataset, output_dataset)
movie_interest = movie_model.predict([[9, 2], [35, 0]])
```

```python
movie_interest
```

```python
from sklearn.svm import SVC
movie_model = SVC()
movie_model.fit(input_dataset, output_dataset)
movie_interest = movie_model.predict([[9, 2], [35, 0]])
```

```python
movie_interest
```

```python
import pickle
# Save your trained model
pickle.dump(movie_model, open('movie_model.pkl', 'wb'))
```
